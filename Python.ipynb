Q8. What are logical operators in Python?

Logical operators combine conditions:

and → True if both conditions are True.

or → True if at least one condition is True.

not → Reverses the result (True → False).



Q9. What is type casting in Python?

Type casting means changing a value from one data type to another, e.g., int("5").



Q10. What is the difference between implicit and explicit type casting?

Implicit: Python converts data type automatically.

Example: 5 + 2.0 → 7.0.

Explicit: You convert manually using functions like int(), float().



Q11. What is the purpose of conditional statements in Python?

Conditional statements let Python make decisions and run certain code only if a condition is True.



Q12. How does the elif statement work?

elif checks another condition if the first if condition is False.

Q13. What is the difference between for and while loops?

for loop: Runs for a fixed number of times.

while loop: Runs as long as a condition is True.



Q14. Describe a scenario where a while loop is more suitable than a for loop.

Q14. Write a program to reverse a string using a while loop.

```python
text = input("Enter a string: ")
reversed_text = ""
i = len(text) - 1
while i >= 0:
    reversed_text += text[i]
    i -= 1
print("Reversed string:", reversed_text)
```

Q15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
num = int(input("Enter a number: "))
fact = 1
while num > 0:
    fact *= num
    num -= 1
print("Factorial:", fact)
```
